**This notebook is for showing data correlation and how to remove them**

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
# Generate sample dataset
np.random.seed(42)
data = {
    'Feature_A': np.random.rand(100) * 10,  # Random values between 0 and 10
    'Feature_B': np.random.rand(100) * 20,  # Random values between 0 and 20
    'Feature_C': np.random.rand(100) * 5    # Random values between 0 and 5
}
df = pd.DataFrame(data)
df.head()
```

**we are giong to introduce some correlation in data**

```python
# Introduce some correlation
df['Feature_B'] = df['Feature_A'] * 1.5 + np.random.rand(100) * 5
df.head()
```

**Compute correlation matrix**

```python
correlation_matrix = df.corr()
print("Correlation Matrix:\n", correlation_matrix)
```

as we can see in the correlation matrix feature_a and feature_b has high postive correlation (0.945204)

**Plot correlation heatmap**

```python
plt.figure(figsize=(8, 6))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt='.2f', linewidths=0.5)
plt.title("Feature Correlation Heatmap")
plt.show()
```

**Scatter plot for highly correlated features**

```python
plt.figure(figsize=(6, 4))
plt.scatter(df['Feature_A'], df['Feature_B'], alpha=0.7, color='b')
plt.xlabel("Feature A")
plt.ylabel("Feature B")
plt.title("Scatter Plot of Feature A vs Feature B")
plt.grid()
plt.show()
```
