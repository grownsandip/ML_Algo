```python
import pandas as pd
path="/content/drive/MyDrive/dataset/Cancer_data.csv"
data=pd.read_csv(path)
```

```python
data.head()
```

```python
data.info()
```

```python
data.drop(["Unnamed: 32","id"],axis=1,inplace = True)
```

```python
data.head()
```

```python
data.diagnosis = [1 if each == "M" else 0 for each in data.diagnosis]
data.info()
```

```python
data.tail()
```

```python
percentage_missing=(100*data.isnull().sum()/len(data))
print("The percentage of missing values:",percentage_missing)
```

```python
y_data=data.diagnosis.values
print(y_data)
pd.DataFrame(y_data)
```

```python
x_data=pd.DataFrame(data)
print(x_data)
```

```python
x_data.head()
```

```python
x_data.info()
```

```python
x = (x_data - x_data.min()) / (x_data.max() - x_data.min())
pd.DataFrame(x)
```

```python
#train test split
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x,y_data,test_size=0.2,random_state=42)
```

```python
from sklearn.linear_model import LogisticRegression
model=LogisticRegression()
model.fit(x_train,y_train)
```

```python
pred=model.predict(x_test)
```

```python
from sklearn.metrics import accuracy_score
accuracy_score(y_test,pred)
```

```python
from sklearn.metrics import confusion_matrix
confusion_matrix(y_test,pred)
```
